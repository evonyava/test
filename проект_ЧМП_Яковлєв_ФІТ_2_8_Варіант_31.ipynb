```python
import pandas as pd

from google.colab import drive

drive.mount('/content/drive')


file_path = '/content/drive/My Drive/Kalush_simpson.xlsx'

df = pd.read_excel(file_path)
df


import numpy as np
import matplotlib.pyplot as plt
plt.figure(figsize=(8, 4))
plt.plot(df['X'], df['Y'], marker='o', linestyle='-', color='b')

plt.title('Візуалізація частини кривої, що обмежує площу району')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)

plt.show()


from scipy.integrate import simpson

def Kalush_simpson(df):
  h = df['X'].iloc[1] - df['X'].iloc[0]
  s = df['Y'].iloc[0] + df['Y'].iloc[len(df)-1]
  for i in range(1, len(df)-1):
    if i % 2 == 0:
      s += 2 * df['Y'].iloc[i]
    else:
      s += 4 * df['Y'].iloc[i]
  return s * h / 3

result_custom = Kalush_simpson(df)

result_scipy = simpson(df['Y'], df['X'])

print(f"Результат власної реалізації: {result_custom}")
print(f"Результат за допомогою simpson: {result_scipy}")
```

```python
file_path1 = '/content/drive/My Drive/Kalush_trapezoid.xlsx'

df1 = pd.read_excel(file_path1)
df1


import numpy as np
import matplotlib.pyplot as plt
plt.figure(figsize=(8, 4))
plt.plot(df1['X'], df1['Y'], marker='o', linestyle='-', color='b')

plt.title('Візуалізація частини кривої, що обмежує площу району')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)

plt.show()


from numpy import trapz

def Kalush_trapezoid(df1):
  integral = 0
  for i in range(1, len(df1)):
        h = df1['X'][i] - df1['X'][i-1]
        integral += (df1['Y'][i] + df1['Y'][i-1]) * h / 2
  return integral

result_custom = Kalush_trapezoid(df1)

result_numpy = trapz(df1['Y'], df1['X'])

print(f"Результат власної реалізації: {result_custom}")
print(f"Результат за допомогою numpy.trapz: {result_numpy}")
```

```python
import pandas as pd
from google.colab import drive
drive.mount('/content/drive')

file_path2 = '/content/drive/My Drive/Kalush_pr.xlsx'

df2 = pd.read_excel(file_path2)
df2


import matplotlib.pyplot as plt

plt.figure(figsize=(8, 4))
plt.plot(df2['X'], df2['Y'], marker='o', linestyle='-', color='b')

plt.title('Візуалізація частини кривої, що обмежує площу району')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)

plt.show()


x_values = df2['X']
y_values = df2['Y']

integral = 0
n = len(x_values) - 1

for i in range(n):
  h = x_values[i+1] - x_values[i]
  f_avg = (y_values[i] + y_values[i+1]) / 2
  integral += h * f_avg

print("Метод середніх прямокутників")
print(f"{integral}")

from scipy.integrate import simpson
result_scipy = simpson(y=df2['Y'], x=df2['X'])

print("Перевірка методом Сімпсона")
print(f"{result_scipy}")
```
