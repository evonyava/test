```python
import numpy as np
import matplotlib.pyplot as plt

def f_a(x, y):
    return x + np.cos(y / np.sqrt(5))

def f_b(x, y):
    return x + np.sin(y / 3)

def euler_method(f, x0, y0, h, a, b):
    n = int((b - a) / h)
    x = np.linspace(a, b, n+1)
    y = np.zeros(n+1)
    y[0] = y0
    for i in range(n):
        y[i+1] = y[i] + h * f(x[i], y[i])
    return x, y

def euler_cauchy_method(f, x0, y0, h, a, b):
    n = int((b - a) / h)
    x = np.linspace(a, b, n+1)
    y = np.zeros(n+1)
    y[0] = y0
    for i in range(n):
        y_pred = y[i] + h * f(x[i], y[i])
        y[i+1] = y[i] + (h/2) * (f(x[i], y[i]) + f(x[i+1], y_pred))
    return x, y

x0_a, y0_a = 1.8, 2.6
a_a, b_a = 1.8, 2.8
h = 0.1

x0_b, y0_b = 1.6, 4.6
a_b, b_b = 1.6, 2.6

x_euler_a, y_euler_a = euler_method(f_a, x0_a, y0_a, h, a_a, b_a)
x_euler_cauchy_a, y_euler_cauchy_a = euler_cauchy_method(f_a, x0_a, y0_a, h, a_a, b_a)

x_euler_b, y_euler_b = euler_method(f_b, x0_b, y0_b, h, a_b, b_b)
x_euler_cauchy_b, y_euler_cauchy_b = euler_cauchy_method(f_b, x0_b, y0_b, h, a_b, b_b)

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(x_euler_a, y_euler_a, 'bo-', label="Euler's Method")
plt.plot(x_euler_cauchy_a, y_euler_cauchy_a, 'r*-', label="Euler-Cauchy Method")
plt.title("Розв'язок для варіанту 1(a)")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(x_euler_b, y_euler_b, 'bo-', label="Euler's Method")
plt.plot(x_euler_cauchy_b, y_euler_cauchy_b, 'r*-', label="Euler-Cauchy Method")
plt.title("Розв'язок для варіанту 1(b)")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()

plt.tight_layout()
plt.show()
```
